```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash", #gemma-4-3lb-it
    temperature=1.0,  # Gemini 3.0+ defaults to 1.0
    max_tokens=None,
    timeout=None,
    max_retries=2,
)

response = model.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
# if you want response in one single line, there is something called system instruction where we can mention that

response = model.invoke([
            ["system","You are a helpful assistant that answers in short one line"],
            ["human","How many moons does Jupiter Have?"]
])

print(response.text)
```

## Call LLM Using Groq Models

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3.6-27b",reasoning_format = "parsed") #here reasoning_Format = "raw" gives complete reasoning process, how it came it to that answer, while "parsed" gives the answer

messages = [
    (
        "system","You are a helpful assistant that answers in short one line"
    ),
    (
        "human","How many moons does Jupiter Have?"
    )
]

ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
