        House Size | No. of Rooms | Price

Here Price is output or dependent variable and other 2 columns are input features.

A)

In ANN, we give features f1 and f2 etc and initiliazed with some weights. Each feature is connected to each node in hidden layer with weights W1, W2, W2.. W6 (2 features connected to 3 nodes in hidden layer with total of 6 weights). Each node of hidden layer is again connected to the output feature with wights W7, W8, W9 ( three weights because three nodes of hidden layer is connected to one output feature ). Finally we get y^. 

Based on these particular weights, we do forward and backward propogation. 
In forward propogation, we take input multiply with weights and we add a bias ( B1, B2 and B3 by each node in hidden layer ) and then we apply a activation function. Finally we get y^. Now, we check loss, which is difference between yi and yi^. If difference is high, we need to reduce the difference and that is done by reducing the weights by backward propogation.

If i interchange or change order of feature from f1, f2 to f2, f1, even then we should be able to train the ANN. that is, the sequence of data doesnt matter here while training the model. Because, at a time, entire row would be sent to train the model and so no matter the order be the data in the rown, they would be all together going to train model.

Problem statement: Lets say we have a data set { sentimental analysis }

            Text                    o/p
            The food is good        1
            The food is bad         0
            The food is not good    0

Lets say, we have all these sentences or documents and ouputs in training set. How to solve this using ANN ?

Firstly, we have to do Text pre-Processing. Text to vectors - BOW, OHE, TFIDF

Vocabulary - 4 unique word ( considering "is" and "the" as stop words ) - good, food, not, bad

Using BOW:

            food good bad not
        S1   1     1    0   0
        S2   1     0    1   0
        S3   1     1    0   1

We have fixed vector size here. 

Now, when creating ANN, we are giving these vector inputs to Neural network. Here, these are text data which means Sequence of data is important. But in vocabulary we considered, the words are not in sequence as compared to sequence of those words in each input Sentences.
So, here when training ANN, we are actually using the sequence of data. When sequence is lost, meaning of sentence is lost because context info is lost. Even with TFIDF or Word2Vec we might lose the sequence.

in sequential data, for eg. "The food is good" -> we have to give input to model at different timestamps in order, first at t=1, The goes in, next at t=2, food goes in, next is goes in and next good goes in at t=4.

But in ANN, all feature data in vector is passed together ( each row is passed together without any importance to sequence ) and then we do forward and backward propogation to minimize the loss. Because of this, Neural network wont be able to train themselves efficiently. 

In case of sequential data, it is always better that we pass the each word in sequnce of time stamps.

So, in case of google translation, where we have to give word after word in sequence, output depends on the sequence of words in input so that a context is maintained.

To solve this, we are going for simple RNN ( Recurrent Neural network )

in RNN, the data would be sequential.

In case of ANN, we will be having input, hidden layer and output layer. When we give input, some processing happens in Hidden layer before ouput is generated.

Suppose there are two inputs, three neurons in hidden layer, and one output. Each feature is connected to each neuron in hidden layer and each neuron in hidden layer is connected to the output. 


In case of RNN, we will be having input, hidden layer and ouput layer. Along with these we will be having a feedback loop. Whatever ouput we are getting from hidden layer, we will be sent to that particular neuron and all other neurons in the hidden layer.

Suppose there are two inputs, three neurons in hidden layer, and one ouput. Each input feature is connected to each neuron in hidden layer. Also, each neuron in hidden layer is connected to the ouput. Until here, it is same as ANN. In addition to this, each neuron in hidden layer will be sending a feeback to that same neuron and other neurons in the hidden layer.

In short, whatever ouput we are getting after we take the inputs along with the weights that are initialized, adding bias and then activation function, Whatever output we are basically getting, we are also sharing with all other neurons in the hidden layer. 

wrt this neural network,

eg. "The food is good" is input, we will be doing numbering based on sentence and word. Here, this is part of first sentence so

The - x11 - first sentence first word
food - x12 - fiest sentence second word
is - x13 - first sentence third word
good - x14 - first sentence fourth word

at time t=1, we would be passing x11 as input, at t=2, we would be passing x12 as input by then the neurons would be already having context of x11 and so on, at t=4 we would be passing x14 as input by then the neurons would be having context of all three earlier words.

So, in case of ANN, we would be passing whole sentence at a time as input but in RNN, we would be sending each word at a time.
