### Importing Modules in Python: Modules and Packages

```python
import math
math.sqrt(10)
```

```python
from math import sqrt,pi

print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np # package to deal with arrays

np.array([1,2,3,4])
```

```python
from math import *

print(sqrt(16))
print(pi)
```

```python
from package.maths import addition
addition(1,3)
```

```python
from package import maths
maths.addition(2,4)
```
