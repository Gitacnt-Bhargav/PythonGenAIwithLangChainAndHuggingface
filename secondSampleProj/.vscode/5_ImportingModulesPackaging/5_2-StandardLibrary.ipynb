```python
## Data serialization

import json

data = {'name':'Krish',"age":25}

json_str = json.dumps(data)  # to get json in form of string
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str) # to convert back to dictionary
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv',mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Krish1','32'])


with open('example.csv',mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# datetime

from datetime import datetime,timedelta

now = datetime.now()
print(now)

yest = now-timedelta(days=1)
print(yest)
```

```python
## time - can be used for program to sleep for few seconds

import time

print(time.time())
time.sleep(2)
print(time.time())
```

```python
## regular expression

import re

pattern = '\d+'
text = 'There are 123 apples out of 200'

match = re.search(pattern,text)
print(match.group()) ## will give 123, the first match in the text
```
