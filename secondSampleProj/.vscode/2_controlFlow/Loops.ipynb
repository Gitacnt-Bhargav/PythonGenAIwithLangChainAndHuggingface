```python
## while loop

count = 0


while count<5:
    print(count)
    count=count+1
```

```python
# Loop control statements

#break

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
# continue

for i in range(10):
    if i%2==0:
        continue
    print(i)
```

```python
## pass - is a null statement that does nothing

for i in range(5):
    if i==3:    
        pass
    print(i)
```

```python
# prime numbes from 1 to 100

for i in range(1,101):
    if i>1:
        for j in range(2,i):
            # print(f"i is {i} and j is {j}")
            if i%j==0:
                break
        else:
            print(i)    

# How it works: In Python, an else attached to a for loop only runs if the loop finishes naturally 
# (i.e., it never hit the break statement).

# Result: It checks every number j. If no number divides i perfectly, the loop finishes, 
# and it prints i. This correctly identifies prime numbers.
```
