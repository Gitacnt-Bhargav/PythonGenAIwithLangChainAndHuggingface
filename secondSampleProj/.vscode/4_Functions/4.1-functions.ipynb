# Functions in Python

```python
# A function is a block of code that performs a specific task

# Syntax
def function_name(parrameters):
    """DocString"""
    #Function Body
    return expression
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
```

```python
even_or_odd(20)
```

```python
## Function with multiple parameters

def add(a,b):
    """to add 2 numbers"""
    return a+b

print(add(1,2))
```

```python
## default parameters

def greet(name):
    print(f"Hello {name}")

greet("Krish")
```

```python
## default parameters

def greet(name="Guest"):
    print(f"Hello {name}")

greet()
greet("Krish")
```

```python
## Variable length arguments
## Positional and keyword argument


# Positional argument
def print_numbers(*args): # universal way of using is *args, but *anyname can be used
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6)
```

```python
# keyword argument

def print_detail(**kwargs): ## in keyword argument, all parameters would be in form of key-value pairs
    for key,value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_detail(name="Krish",age="32",country="India")
```

```python
def print_details(*args,**kwargs):
    for val in args:
        print(f"Positional argument:  {val}")
    
    for key,value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_details(1,2,3,"Krish",name="krish")
```

```python
print_details(1,2,3,name="krish",4)
```

```python
### Return statement

def multiple(a,b):
    return a*b

multiple(10,4)
```

```python
## return multiple parameters from a function

def multiple(a,b):
    return a*b,a

multiple(2,3) # will return a tuple of 2 values
```
