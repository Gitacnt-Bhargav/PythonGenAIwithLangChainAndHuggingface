```python
#Example 1 : Temperature Conversion

def convert_temperature(temp, unit):
    """This function converts temperature between Celsius and Fahrenhiet"""
    if unit=='C':
        return temp * 9/5 +32 # Celsius to Fahrenhiet
    elif unit=='F':
        return (temp-32)*5/9 # Fahrenhiet to Celsius
    else:
        return None

print(convert_temperature(25,'C'))
print(convert_temperature(77,'F'))
```

```python
#Example 2 : Password strength checker

def is_strong_password(password):
    """This function checks if the password is stron or not"""
    if len(password)<8:
        return False,"Is less than 8 characters"
    if not any(char.isdigit() for char in password):
        return False,"Doesnt contain any digit"
    if not any(char.islower() for char in password):
        return False,"Doesnt contain any lower case letter"
    if not any(char.isupper() for char in password):
        return False,"Doesnt contain any upper case letter"
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False,"Doesnt contain any special character"
    return True

print(is_strong_password("hello"))
print(is_strong_password("a*B123ll"))
```

```python
#Example 3: Calcilate the total cost of items in a shopping cart

def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    return total_cost

## Example cart data
cart=[
    {'name':'Apple','price':0.5,'quantity':4},
    {'name':'Banana','price':0.3,'quantity':6},
    {'name':'Orange','price':0.7,'quantity':3}
]

print(calculate_total_cost(cart))
```

```python
#Example4: Check if a string is palindrome

def is_palindrome(s):
    str = s.lower().replace(" ","")
    if str==str[::-1]:
        return f"{str} Is a palindrome"
    else:
        return f"{str} Is not a palindrome"
    
print(is_palindrome("aha"))
print(is_palindrome("Hello"))
print(is_palindrome("A man a plan a canal Panama"))
```

```python
#Example5: Calculate the factorials of a number using recursion



def factorial(n):
    if n==0:
        return 1
    elif n==1:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(3))
print(factorial(10))
```

```python
#Example6: Calculate the fibonacci of a number using recursion in more faster way

from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):
    if n<2:
        return n
    return fibonacci(n-1)+fibonacci(n-2)

print(fibonacci(3))
print(fibonacci(10))
```

```python
#Example 7: Function to read a text from a file and count frequency of each word

def count_word_frequency(file_path):
    word_count={}
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('./!?:"\'')
                word_count[word] = word_count.get(word,0)+1
    
    return word_count

file_path = 'sample.txt'
word_freq = count_word_frequency(file_path)
print(word_freq)
```

```python
#Example 7: Validate Email address

import re

#Email validation function
def is_valid_email(email):
    """This function checks if the email is valid"""
    pattern1=r'^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9-]+\.[a-zA-z0-9-.]+$'
    pattern2=r'^[a-zA-Z0-9_.+-]+@gmail.com$'
    return (re.match(pattern1,email) or re.match(pattern2,email)) is not None

print(is_valid_email("test@example.com"))
print(is_valid_email("invalid-mail"))
print(is_valid_email("abc+1@gmail.com"))
```
