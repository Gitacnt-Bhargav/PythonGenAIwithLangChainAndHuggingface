## The map() Function in Python

The map() function applies a given function to all items in an input list ( or any other iterable ) and returns a map object (an iterator). This is particularly useful for transforming data in a list comprehensively.

Map takes two inputs 1- a function and 2- iterable like list

```python
numbers = [1,2,3,4,5,6,7,8]

list(map(lambda x:x**2,numbers))
```

```python
def square(num):
    return num**2

numbers = [1,2,3,4,5,6,7,8]

list(map(square,numbers))
```

```python
# Map multiple iterables

numbers1 = [1,2,3]
numbers2 = [4,5,6]

added_numbers = list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```

```python
## map() to covert a list of strings to integers

strs = ["krish","naik","hello","world"]

vals = list(map(lambda str:len(str),strs))
print(vals)
```

```python
# map() to covert to upper case

strs = ['apple','banana','orange']

upper_words = list(map(str.upper,strs))
print(upper_words)
```

```python
def get_name(person):
    return person['name']

people = [
    {'name':'Krish','age':32},
    {'name':'Jack','age':34}
]

list(map(get_name,people))
```
