```python
# Tuples

# Tuples are ordered collection of items that are immutable. They are similar to lists but their immutability 
# makes them diff
```

```python
## creating a tuple

empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))

tpl = tuple()
print(type(tpl))

```

```python
numbers = tuple([1,2,3,4,5]) # converted from list to tuple
print(numbers)
```

```python
list((1,2,3,4,5)) # converted from tuple to list
```

```python
mixed_tuple = (1,"Hello world",3.14,True)

print(mixed_tuple)
```

```python
## Accessing tuple elements

numbers = tuple([1,2,3,4,5])

print(numbers[1])
print(numbers[-2])
print(numbers[2:])
print(numbers[:2])
print(numbers[::2])
print(numbers[::-1])
```

```python
## Tuple Operations

concatenation_tuple = numbers + mixed_tuple

print(concatenation_tuple)
```

```python
mixed_tuple * 3
```

```python
numbers * 3
```

```python
## Immutable nature of tuples

lst = [1,2,3,4,5]
print(lst)

lst[1] = "Krish"
print(lst)
```

```python
numbers = tuple([1,2,3,4,5])
numbers[2] = "Krish"
```

```python
# Tuple methods

numbers = tuple([1,2,3,4,5,1,1])

print(numbers.count(1))
print(numbers.index(3)) # returns first index of value
```

```python
# Packing and Unpacking tuple

# Packing
packed_tuple =  1,"Hello",3.14
print(packed_tuple)
```

```python
# unpacking a tuple

a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
# Unpacking with *

numbers = (1,2,3,4,5)
first,*middle,last = numbers

print(first)
print(middle)
print(last)
```

```python
## Nested tuple

lst = [[1,2,3,4],[4,5,6,7],[1,"Hello",3.14,"c"]]
print(lst[0])
print(lst[1][2])
print(lst[0][:2])
```

```python
lst = [[1,2,3,4],[4,5,6,7],(1,"Hello",3.14,"c")] # tuple inside a list

lst[2][0:3] 
```

```python
nested_tuple = ((1,2,3,4),("a","b","c"),(True,False))

## access the elements inside tuple
print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
# iterating over nested tuples

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end = " ")
    print()
```
