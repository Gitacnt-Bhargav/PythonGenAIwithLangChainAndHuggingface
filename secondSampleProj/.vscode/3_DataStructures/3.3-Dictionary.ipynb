# Dictionary

Dictionaries are unordered collection of items. They store data in key-value pairs. Keys must be unique and immutable (eg. strings, numbers, or tuples), while values can be of any type

```python
## Creating dictionaries

empty_dictionary = {}
print(type(empty_dictionary))
```

```python
empty_dict = dict()
print(empty_dict)
print(type(empty_dict))
```

```python
student = {"name" : "Krish", "age":32, "grade":24}
print(student)
print(type(student))
```

```python
#Keys are unique
student = {"name" : "Krish", "age":32, "name":24}
print(student)
```

```python
 #accessing Dictionary elements
student = {"name" : "Krish", "age":32, "grade":"A"}
print(student)

print(student['grade'])
print(student["age"])
print(student.get('grade'))
print(student.get("name"))

#get value for key last_name if found, otherwise set default value as the 2nd param
print(student.get("last_name","Not Available")) 
```

```python
# Modifying Dictionary elements - Dictionary are mutable, so you can add, update or delete elements

print(student)
```

```python
student['age']=33 #updated value for the key
print(student)
student["address"] = "India" #added a new key and value
print(student)
del student['grade']
print(student)
```

```python
# Dictionary methods

#to get all keys
keys = student.keys()
print(keys) # prints list of keys

values = student.values()
print(values) # prints list of values

items = student.items()
print(items) #prints all key value pairs that are stored as list of tuples
```

```python
# Shallow copy

student_copy =  student

print(student)
print(student_copy)
```

```python
student["name"] = "Krish2"
print(student)
print(student_copy)

# changes in one dictionary has updated similarly in other dictionary as well due to copy
```

```python
student_copy1 = student.copy()
print(student)
print(student_copy1)
```

```python
student["name"] = "Krish3"
print(student)
print(student_copy1)
```

```python

# iterating over dictionaries

# You can use loops to iterate over dictionaries - keys, values, or items

## iterating over keys
for keys in student.keys():
    print(keys)
```

```python
# iterate over values
for values in student.values():
    print(values)
```

```python
#iterate over key value paies
for key,value in student.items():
    print(f"{key} : {value}")
```

```python
# Nested dictionaries

students = {
    "student1":{"name":"Krish","age":32},
    "student2":{"name":"Peter","age":35}
}

print(students)
```

```python
## Access nested dictionaries elements

print(students["student1"]["name"])
print(students["student2"]["age"])
```

```python
## iterating over nested dictionaries

for student_id,student_info in students.items():
    print(f"{student_id} : {student_info}")
    for key,value in student_info.items():
        print(f"{key} : {value}")
```

```python
# Dictionary comprehension

squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
# Conditional Dictionary comprehension

evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

```python
## Practical Examples

## Use a dictionary to count frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4,4]

# using dictionary comprehension
frequency = {x:numbers.count(x) for x in numbers}
print(frequency)

# using for loop

freq = {}
for number in numbers:
    freq[number] = freq.get(number,0)+ 1

print(freq)

## or

freqs = {}
for number in numbers:
    if number in freqs:
        freqs[number]+=1
    else:
        freqs[number]=1

print(freqs)
```

```python
# Merge 2 dictionaries into one

dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}

merged_dict= {**dict1,**dict2} ## ** means, any number of key value pairs is just gonna append
## both dict1 and dict2 are given as keyword arguments

print(merged_dict)
```
