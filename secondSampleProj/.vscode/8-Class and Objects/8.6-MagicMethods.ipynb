### Magic methods

Magic method in Python, also known as dunder methods (double underscore methods), are special methods
that can start and end with double underscores. These methods enable you to define the behavior of objects for built-in operations, such as arithmetic operations, comparisons, and more

Magic methods are predeined methods in Py that you can override to change the behabior of your objects. Some magic methods:

- init: intializes a new instance of a class
- str: Returns a string representation of an object
- repr: Returns an official string representation of an object
- len: Returns length of an object
- getitem: Gets an item from a container.
- setitem: Sets an item in a container

```python
## Basic methods

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age    

person = Person("Krish",34)
print(person)
print(repr(person))
```

```python
## Basic methods

class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name} is {self.age} years old"    
    
    def __repr__(self):
        return f"Person(name={self.name}, age = {self.age})"

person = Person("Krish",34)
print(person)
print(repr(person))
```
