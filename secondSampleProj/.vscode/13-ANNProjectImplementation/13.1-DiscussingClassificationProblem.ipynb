

libraries: Tensor flow , Keras

Tensor Flow - Open source library specifically used to create E2E deep learning projects. Can implement ANN to RNN neural networks, LSTM RNN and GRU RNN

If only using Tensor flow, then we have to write lot of code but using Keras, which is again integrated into Tensor flow, provides APIs which help in creating Neural networks with less lines of code


We use ANN to solve, if we have a tabular data, when we have some classification or regression problem statement, we would be able to solve with help of ANN.

We will be working with Churn_Modelling.csv where data is from a bank, where based on some info it says whether the customer is about to leave the bank or not.
So the dataset has past data, where some data for customer were there and whether he has left the bank.

Here, we will be solving Simple binary classification problem with help of ANN.

Process:

Dataset: Churn Modelling
Focus on: Classification

Initially we would be taking the data set and perform some basic Feature engineering - convert categorical variables into Numerical, use some standardization process

Here, we have 11 independent features and so input layer would have 11 nodes. We would be having hidden layer. There would be one output layer with one output. Since its ANN, each node in input layer would be connected to every node in hidden layer.

We will use Drop out - which means, some of the nodes will be disabled using forward propogation and backward propogation, we wont be updating some of the weights. This is to make sure that our model doesnt overfit.

After training the model, we will see the concept of Loss, Optimizers all with help of Keras and Tensor flow. 

Once the mdoel is trained, how do we take the model and convert into some file format like pickle file format or .h5 format. This will be useful for deployment and can be reused in any other environment

Further we will use Streamlit to create our webapp and integrate all our models and will deploy into streamlit cloud
