### File operation - Read and write files

```python
## Read a whole file

with open('example.txt',mode='r') as file:
    content = file.read()
    print(content)
```

```python
## read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line)
```

```python
## read a file line by line

with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) # strip() removes the newline character
```

```python
## writing a file (overwriting)

with open('example.txt','w') as file: # write mode
    file.write('Hello world!\n')
    file.write("this is a new line.")

# the entire content is overwritten 
```

```python
## write a file (without overwriting)

with open('example.txt','a') as file: # append mode
    file.write("\nAppend operation taking place")
```

```python
## writing a list of lines to a file

lines = ['First line\n','Second line\n','Third line\n']

with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## binary files (.bin files)

data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file: # write byte mode
    file.write(data)
```

```python
## read binary file

with open('example.bin','rb') as file: # write byte mode
    content = file.read()
    print(content)
```

```python
### read content from a source text file and write a destination text file

with open('example.txt','r') as source_file:
    content = source_file.read()

with open('destination.txt','w') as dest_file:
    dest_file.write(content)
```

```python
### Read a text file and count the number of lines, words and characters

with open('example.txt','r') as file:
    content = file.read()

print(content)

def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file = 'example.txt'
lines,words,chars = count_text_file(file)
print(f"has {lines} lines, {words} words and {chars} chars")
```

```python
# Writing and then reading a file

# w+ mode in python is used to open a file for both reading and writing. If the file 
#does not exist, it will be created. if the file exists, its content is truncated (i.e., the
# file is overwritten)

with open('example.txt','w+') as file:
    file.write('Hello world\n')
    file.write('This is a new line \n')

    ## move the file cursor to the beginning so as to read the content of the file
    file.seek(0)

    ##read the content of the file
    content = file.read()
    print(content)
```
