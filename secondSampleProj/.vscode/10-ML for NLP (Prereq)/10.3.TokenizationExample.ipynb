```python
corpus = """Hello Welcome, to Krish Naik's NLP Tutorials.
Please do watch the entire course! to become expert in NLP.
"""
```

```python
corpus
```

```python
print(corpus)
```

```python

## Tokenization.

## Paragraph to Sentence

from nltk.tokenize import sent_tokenize # this package converts paragraph to sentences

documents = sent_tokenize(corpus) # this gives list of sentences. it has split with . and !.
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
## Tokenization

## Paragraph to words
## Sentence to words

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
words_from_para = word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
## 

from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus) # here 's is already split rather than single word in case of word_tokenize. wordpunct_tokenize makes sure that
    #punctuation is also converted to a separate word
```

```python
# Tree bank word tokenizer

from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

```python
# with help of Tree bank tokenizer, . (pullstop) in between the corpus is not considered as separate word like as in word_tokenizer 
# or wordpunct_tokenizer

#But the las . would be considered as separate word.

# Most of the times we use word_tokenize and sent_tokenize
```
