### One Hot Encoding

Words to Vector

For eg. consider 3 documents/sentences

D1 - The food is good -> o/p: 1
D2 - The food is bad -> o/p: 0
D3 - Pizza is Amazing -> o/p: 1

Here unique words or Vocabulary - The food is good bad Pizza Amazing

Based on this unique words, One hot encoding will represent data as below vectors
    The food is good bad Pizza Amazing
The: 1   0   0   0   0    0     0
food: 0  1   0   0   0    0     0
.
.
.

Now D1 becomes: [[1000000], [0100000],[0010000],[0001000]] : Shape is (4 X 7)
D2: [[1000000], [0100000],[0010000],[0000100]] : Shape is (4 X 7)
D3: [[0000010], [0010000],[0000001]] : Shape is (3 X 7)

These are One Hot encoded format.

### Advantages and Disadvantages of OHE

#### Advantages

1. Easy to implement with python. In sklearn, we have library - One Hot encoder and with pandas, pd.get_dummies() will help to create OHE


#### Disadvantages

Sparse Matrix - in an array having lots of 1s and 0s. 

1. Sparse Matrix - usually leads to Overfitting. With test data, it gives good accuracy and with any new data, it wont give good accuracy
2. With One Hot encoding - the text size is not same. For above eg., the shape of three sentences are not all same. for 2 it is 4X7 and 1 is 3X7
For ML use case, the number of features should be fixed wrt length. The difference in size for above eg. makes unsuitable to train the ML algorithm
3. No Semantic meaning is getting captured. For eg. Food, Pizza, burger are 3 words in vocabulary - so OHE will be 100, 010, 001. Since there are 3 features, there would be 3 vectors. With OHE, it is hard to differentiate between those words, because of which we are not able to find, which is most important word, which word is more similar to other words etc. 
4. Out of vocabulary (OOV) - In above eg, if test data is "Burger is bad" and is asked to predict, then algotithm wont understand the word Burger since that word was no where in Vocabulary. This means, the Algorithm is not able to deal with Out of Vocabulary words
5. If the vocabulary size is 50K, then the 0s and 1s would be so large leading to Sparse metrics
