### SKIPGRAM - type of Word2Vec

Corpus:

iNeuron Company is related to Data Science

Window Size = 5

o/p                                                                                                                 i/p
iNeuron, Company, Related, to [forward and backword words of that center word of first 5 words]                     IS

Next 5 words where center word is related
o/p                                                                                                                 i/p
Company, Is, To, Data                                                                                               Related

o/p                                                                                                                 i/p
Is, related, Data, Science                                                                                          To

When creating a neural network, input layer would have one word with 7 vectors ( size of vocabulary ), window size vector in the hidden layer, and in ouput there would be 4 words each with 7 vectors. [ This is completely opp to CBOW ]

When i/p goes as IS, then vector 0010000 would go, the expected ouput and the computed output using ANN softmax function is compared and the comparision is done using LOSS function. Continuous Forward an backward propogation would be done to minimize the LOSS function. 



When should we apply CBOW or SkipGram

According to research, when there is a small corpus then can go with CBOW and with large corpus, then go with SkipGram.

> To increase CBOW or SkipGram accuracy, increase the training data. 
> Also, increase in window size, in turn leads to increase of vector dimensions, better performance would be acheived

### Advantages of Word2Vec


1. Previously we used to get Sparse matrix ( having 0s and 1s mostly in matrix ) causing Overfitting. With Word2Vec, we are able to get Dense matrix, there wont be that many 0s and 1s. With Dense Matrix, ML model would be able to get trained in a better way
2. Semantic information is getting capture, with this we would be able to find out similar vectors.
3. No worry about Vocabulary size as like in BOW and TFIDF, where vocabulary size was to be considered. With Word2Vec we will have fixed set of dimension. eg. with Google pretrained Word2Vec, there will be only 300 dimensions
4. Out of vocabulary - There will be hardly any word that is not present in the corpus or vocabulary
