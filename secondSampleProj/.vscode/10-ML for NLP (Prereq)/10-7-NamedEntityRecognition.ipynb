```python
sentence = """The Eiffel tower was build from 1887 to 1889 by French Engineer Gustave Eiffel, 
whose company specialized in building metal frameworks and structures"""
```

```python
import nltk
words = nltk.word_tokenize(sentence)
```

```python
nltk.pos_tag(words)
```

```python
tagged_elements =nltk.pos_tag(words)
```

```python
nltk.download("maxent_ne_chunker_tab")
nltk.download("words")
```

```python
nltk.ne_chunk(tagged_elements).draw()
```
