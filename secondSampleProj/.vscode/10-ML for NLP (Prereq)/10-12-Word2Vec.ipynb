### Word Emberddings

In NLP, word embedding is a term used for the representation of words for text analysis, typically in the form of a real-valued vector that encodes the meaning for the word such that the words that are closer in the vector spaces are expected to be similar in the meaning



eg. We have two words - Happy and Excited. With help of Word Embeddings, convert to Vector and plot this on 2D graph using PCA or any unsupervised way by dimension reduction, then the distance between Happy and Excited is less than as compared to if for one more word Angry.

OHE, BOW, TFIDF are all part of Word Embedding 

Word Embedding is of 2 types:

1. Count or Freq  - OHE, BOW, TDIDF
2. Deep leaening Trained Model - will give high accuracy - Word2Vec


Word2Vec is of 2 types

1. CBOW - Continuous Bag of Words (needs Prereq - how ANN works)
2. Skipgram


### WORD2VEC

Word2Vec is a kind of Word embedding technique where the focus is to convert word into vectors making sure that the meaning of diff words is maintained (vectors of similar words are near)

Word2Vec is a technique for NLP published by Google in 2013 trained on around 3Bn words
The word2vec algorithm uses a neural network model to learn word associations from a large corpus of text
Once trained, such a model can detect synchronous words or suggest additional words for a partial sentence
As the name implies, word2vec represent each distinct word with a particular list of numbers called a vector

Suppose we have a vocabulary - unique words in a Corpus

Feature Representation
-- Each and every word present in the vocabulary will be converted into a Feature representation. This means that, we are going to convert words to vectors based on some features

### Cosine similarity

each vocabulary is represented by 2 vectors:
KING [0.95 0.96], MAN [0.95 0.98], QUEEN [-0.96 0.95], WOMEN [-0.94 -0.96]

When we do calculation : KING - MAN + QUEEN then o/p would be WOMEN


Distance between two vectors plotted gives how similar those words are, and for that formula is:

Distance  = 1 - Cosine Similarity

Cosine Similarity = Angle between two vectors, for eg. if its 45 degrees, then Cos 45 = 1/sqrt(2) = 0.7071

Distance = 1 - 0.7071 = 0.29

Distance if nearer to 0 means they are almost similar or if its almost 1 then they are almost different
