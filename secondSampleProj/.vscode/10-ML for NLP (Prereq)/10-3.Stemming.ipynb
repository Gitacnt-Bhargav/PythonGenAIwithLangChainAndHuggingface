### Stemming

Stemming is the process of reducing a word to its word step that affixes to sufixes and prefixes or to the roots of words known as lemma.

Stemming is important in natural language understanding (NLU) and NLP

This is used in text preprocessing to clean the data

Classification problem

Comments on the product is a postive review or negative review

Reviews ----> eating, eat, eaten,. Here eat is the stem word and all words are basically talking about eat

similarly if go, going, gone etc are words, then go is stem word, rest all words are actually talking about go.

So instead of having similar kind of words, can have one stem word and try to solve the problem wrt that

```python
words = ["eating","eat","eaten","writing","writes","programming","programs","history","finally","finalize"]

## finding stem words
```

#### PorterStemmer

```python
from nltk.stem import PorterStemmer
```

```python
stemming= PorterStemmer()
```

```python
for word in words:
    print(word + "----> " +stemming.stem(word))
```

```python
## here history has changed to histori, changing the complete meaning. This is the major disadvantage of stemming

stemming.stem("congratulations") # is giving "congratul" which is not giving any meaning
```

```python
stemming.stem("sitting") # this works fine
```

### RegexpStemmer class

NLTK has RegexStemmer class with the help of which we can easily implement Regular expression Stemmer algorithms. 
It basically takes a single regex, and removes any prefix or suffix that matches the expression

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4) ## this means, that wherever there is ing, s, e, able at last, remove it
```

```python
reg_stemmer.stem("eating")
```

```python
reg_stemmer.stem("ingeating") # regex only says that remove last, change regex to ing instead of ing$, then all ing would be removed from the 
#word, and to remove only from the start, use ^ing
```

### Snowball Stemmer

better technique compared to PorterStemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballStemmer =  SnowballStemmer("english")
```

```python
for word in words:
    print(word + "---> " + snowballStemmer.stem(word))
```

```python
stemming.stem("fairly"), stemming.stem("sportingly")
```

```python
snowballStemmer.stem("fairly"), snowballStemmer.stem("sportingly")
```

```python
snowballStemmer.stem("going")
```

```python
snowballStemmer.stem("goes")
```

```python
stemming.stem("goes")
```

No matter how improved technique it is, there are still some words that doesnt come out properly. So, Stemming cannot be used for chatbots etc.

For chatbots, we can go with Lemmatization
