### Lemmatization

For few words, Stemming was completely changing meaning and thus this technique is not suitable for chatbots etc

### Wordnet Lemmatizer

Lemmaization technique is like Stemming. The output we will get after Lemmatization is called 'lemma', which is a root words rather than root word stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a Lemma.

Can be used for Chatbots, Q&A, text summarization

```python
import nltk
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
lemmatizer.lemmatize("going")
```

```python
'''
POS:
Noun - n
verb - v
adjective - a
adverb - r
'''
#by default n

lemmatizer.lemmatize("going",pos='v')
```

```python
lemmatizer.lemmatize("going",pos='r')
```

```python
words = ["eating","eat","eaten","writing","writes","programming","programs","history","finally","finalize"]
```

```python
for word in words:
    print(word + "---> " + lemmatizer.lemmatize(word))
```

```python
for word in words:
    print(word + "---> " + lemmatizer.lemmatize(word,'v'))
```

```python
lemmatizer.lemmatize("goes")
```

```python
lemmatizer.lemmatize("fairly"), lemmatizer.lemmatize("sportingly")
```

Wordnet Lemmatizer takes more time compared to Stemming
