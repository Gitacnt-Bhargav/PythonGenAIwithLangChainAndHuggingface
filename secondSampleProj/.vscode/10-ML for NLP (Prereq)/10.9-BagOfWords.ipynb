### Bag of Words

Data set is 

He is a good boy        o/p: 1
She is a good girl      o/p: 1
Boy and girl are good   o/p: 1

Step2: Lower all the words, and apply stop words

after lowering:
he is a good boy
she is a good girl
boy and girl are good

after applying stop words:
good boy
good girl
boy girl good

Vocabulary: 
good   freq: 3
boy    freq: 2
girl   freq: 2

(Order Based on descending order)

Imp of Freq is: If in vocabulary there are 100 words, and there are some words that are present only once, those need not be considered. When coding BOW, we have option to select top 10 or 20 features based on freq. 

Based on top most freq: 
good boy girl

now, senteces would become
                    good    boy     girl
S1 - good boy -     1       1       0
S2 - good girl -    1       0       1
S3 - boy good girl - 1      1       1

There are two things, 

Binary Bag of words, Normal Bag of words.

The difference between these are, suppose in a sentence, the word which is part of vocabulary occured twice in a sentence, then the value to that word is increased to 2 in case of Normal bag of words, but in case of Binary bag of words, no matter the word repeats any number of times, the value would be 1

eg. S1 - good boy good -> 2 1 0 would be value set by Normal BoW and 1 1 0 would be value set by Binary BoW

In binary BoW, there would be only 1s and 0s, while in Normal BoW, count would be updated based on freq of that word in that sentence



### Advantages and Disadvantages with BOW

#### Advantages:

1. Simple and Intuitive
2. Semantic problem is fixed which was there in OHE by fixed Sized input which will be suitable and helpful for ML algorithm training


#### Disadvantages:

1. Sparce Matix or array is still there, which will cause overfitting
2. The ordering of word gets changed and so meaning changes - The vocabulary is ordered in descending order and when BOW is used, the vector created will be not in same order as in sentence.
3. Out of Vocabulary problem is still present, since that word which is of least freq would not be part of vocabulary, Eventhough if the word is important.
4. Semantic meaning is still not getting captured. (as present in OHE). For eg. "The food is good" and another sentence is "The food is not great". The vectors created would be [ 1 1 1 0 1] and [1 1 1 1 1], when this is plotted on 2D plane using PC, both points would be very near since the diff is very less ( only one value is diff ), and so the cosine similarity would high because the angle between two near points would be less. This shows that both sentences are almost similar eventhough both give completely diff meaning.
