### N-grams

Suppose there are 2 sentences
1. The food is good
2. The food is not good

Vocabulary after removing possible stopwords: food not good

S1 - 1 0 1
S2 - 1 1 1

These 2 sentences are almost similar when checking on the vector, since there is only a difference in one value

Now, with N-gram:
initial Vocabulary - food not good
Vocabulary - "food" "not" "good" "food good" "food not" "not good" 
This is after bigram, similarly we can have trigram also..

Now sentences wrt new Vocabulary would become:
    "food" "not" "good" "food good" "food not" "not good" 
S1 - 1 0 1 1 0 0
S2 - 1 1 1 0 1 1

Now, model will be able to distinguish since there is difference between multiple values

in sklearn -> there is a parameter - n-gram, if its value is 

(1,1) - unigram
(1,2) - unigram, bigram
(1,3) - unigram, bigram, trigram
(2,3) - bigram, trigram
