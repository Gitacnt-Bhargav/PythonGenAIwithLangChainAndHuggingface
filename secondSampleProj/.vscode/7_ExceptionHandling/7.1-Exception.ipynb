```python
try:
    num= int(input("Enter the number: "))
    result = 10/num
except ValueError:
    print("This is not a proper number")
except ZeroDivisionError:
    print("enter denominator greater than zero")
except Exception as ex:
    print(ex)
```

```python
# try, except, else block

try:
    num= int(input("Enter the number: "))
    result = 10/num
except ValueError:
    print("This is not a proper number")
except ZeroDivisionError:
    print("enter denominator greater than zero")
except Exception as ex:
    print(ex)
else:
    print(f" The result is {result}") #else block runs when try block works fine without exception
```

```python
# try, except, else and finally

try:
    num= int(input("Enter the number: "))
    result = 10/num
except ValueError:
    print("This is not a proper number")
except ZeroDivisionError:
    print("enter denominator greater than zero")
except Exception as ex:
    print(ex)
else:
    print(f" The result is {result}") #else block runs when try block works fine without exception
finally: # will execute if there is or no exception. will execute anyhow
    print("Execution completed") 
```
