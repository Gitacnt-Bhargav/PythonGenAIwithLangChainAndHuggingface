```python
num1 = 10
num2 = 5

sum = num1 + num2
diff = num1 - num2
product = num1 * num2
quotient = num1 / num2
floor_div = num1 // num2 # gives the floor value removing the decimal
modulus = num1 % num2 # remainder 
exponent = num1 ** num2

print(sum)
print(diff)
print(product)
print(quotient)
print(floor_div)
print(modulus)
print(exponent)
```
