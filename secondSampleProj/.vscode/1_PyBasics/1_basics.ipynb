```python
# Semantics: type inference. during execution,\
#  we can change the type of variable

variable =10
print(type(variable))
variable = "abc"
print(type(variable))
```

```python
age=10
height = 6.1
name = "Bhar"
is_student = True
```

```python
print("age: ",age)
print("height: ",height)
print("name: ",name)
print("is_student: ", is_student)
```

```python
print(type(age))
print(type(height))
print(type(name))
print(type(is_student))
```

```python
age_str = str(age)
print(age_str)
print(type(age_str))
```

```python
## dynamic typing
## python allows type of variable to change as program executes
var = 10
print(var, type(var))

var = "Hello"
print(var, type(var))

var = 3.14
print(var, type(var))
```

```python
## input

age = input("Enter age: ")
print(age,type(age))
```

```python
age = int(input("Enter age again: "))
print(age,type(age))
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

sum = num1 + num2
diff = num1 - num2
product = num1 * num2
quotient = num1 / num2

print(sum, type(sum))
print(diff, type(diff))
print(product, type(product))
print(quotient, type(quotient))
```
